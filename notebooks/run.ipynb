```python
import kagglehub
import shutil
from pathlib import Path

# 1) ดาวน์โหลด (จะได้ path ที่เก็บไฟล์ในเครื่อง)
path = kagglehub.dataset_download("nih-chest-xrays/data")
print("Downloaded to:", path)

# 2) ย้าย/คัดลอกไปไว้ในโปรเจกต์ของเรา: data/nih
project_data_dir = Path("../data/nih").resolve()
project_data_dir.mkdir(parents=True, exist_ok=True)

# คัดลอกทั้งโฟลเดอร์ dataset มาไว้ที่ data/nih
shutil.copytree(path, project_data_dir, dirs_exist_ok=True)

print("Copied to:", project_data_dir)
```
